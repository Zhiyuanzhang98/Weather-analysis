```python
data = pd.read_csv('all_data.csv')
```

```python
# 确保日期格式正确
data['日期'] = pd.to_datetime(data['日期'])

# 检查数据的基本信息
print(data.info())
print(data.describe())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS']
# 分城市绘制 AQI 时间序列
plt.figure(figsize=(10, 3),dpi=200)
for city in data['loca'].unique():
    city_data = data[data['loca'] == city]
    plt.plot(city_data['日期'], city_data['AQI指数'], label=city)

plt.title('城市 AQI 指数时间序列')
plt.xlabel('日期')
plt.ylabel('AQI 指数')
plt.legend()
plt.show()
```

```python
# 空气质量等级分布
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
plt.figure(figsize=(10, 3),dpi=200)
quality_counts = data.groupby(['loca', '质量等级']).size().unstack()
quality_counts.plot(kind='bar', stacked=True)

plt.title('不同城市空气质量等级分布')
plt.xlabel('城市')
plt.ylabel('天数')
plt.show()
```

```python
# PM2.5 和 PM10 平均值对比
pollutants = data.groupby('loca')[['PM2.5', 'PM10']].mean()
plt.figure(figsize=(10, 3),dpi=200)
pollutants.plot(kind='bar')
plt.title('各城市 PM2.5 和 PM10 浓度对比')
plt.ylabel('浓度 (µg/m³)')
plt.show()
```

```python
# 特征相关性热力图
corr = data[['PM2.5', 'PM10', 'So2', 'No2', 'Co', 'O3']].corr()

print(corr)

plt.figure(figsize=(5, 4),dpi=200)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('污染物相关性热力图')
plt.show()
```

```python
# AQI 排名分布
plt.figure(figsize=(6, 2),dpi=200)
sns.histplot(data=data, x='当天AQI排名', hue='loca', bins=30, kde=True)

plt.title('北京和宁波 AQI 日排名分布')
plt.xlabel('AQI 排名')
plt.ylabel('频率')
plt.show()
```

```python
for city in data['loca'].unique():
    plt.figure(figsize=(20, 5),dpi=200)
    city_data = data[data['loca'] == city]
    pivot = city_data.pivot_table(index='日期', columns='质量等级', values='AQI指数', aggfunc='count').fillna(0)
    pivot.plot.area(stacked=True, alpha=0.6, title=f'{city} 空气质量等级时间分布')
    plt.ylabel('天数')
    plt.xlabel('日期')
    plt.show()
```

```python
# 添加季节列
def get_season(month):
    if month in [3, 4, 5]:
        return '春季'
    elif month in [6, 7, 8]:
        return '夏季'
    elif month in [9, 10, 11]:
        return '秋季'
    else:
        return '冬季'

data['季节'] = data['日期'].dt.month.apply(get_season)

# 季节性 AQI 平均值
seasonal_aqi = data.groupby(['loca', '季节'])['AQI指数'].mean().unstack()

seasonal_aqi.plot(kind='bar', figsize=(10, 6), colormap='viridis')
plt.title('各城市四季 AQI 平均值对比')
plt.ylabel('AQI 指数')
plt.xlabel('城市')
plt.legend(title='季节')
plt.show()
```

```python
data['月份'] = data['日期'].dt.month

# 按月绘制箱线图
plt.figure(figsize=(6, 3), dpi=200)
sns.boxplot(data=data, x='月份', y='PM2.5', hue='loca')

plt.title('各城市每月 PM2.5 浓度分布')
plt.xlabel('月份')
plt.ylabel('PM2.5 浓度 (µg/m³)')
plt.legend(title='城市')
plt.show()
```

```python
plt.figure(figsize=(10, 3), dpi=200)
for city in data['loca'].unique():
    city_data = data[data['loca'] == city]
    plt.plot(city_data['日期'], city_data['PM2.5'].rolling(7).mean(), label=f'{city} PM2.5 (7日滑动平均)')

plt.title('每日 PM2.5 浓度波动趋势')
plt.xlabel('日期')
plt.ylabel('PM2.5 浓度 (µg/m³)')
plt.legend()
plt.show()
```

## 机器学习

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.linear_model import LinearRegression
```

```python
data = bj_data1824
```

```python

data_bj = bj_data1824
data_bj['日期'] = pd.to_datetime(data_bj['日期'])
data_bj['month'] = data_bj['日期'].dt.month
data_bj['day'] = data_bj['日期'].dt.day
data_bj = data_bj.drop(['日期', '质量等级'], axis=1)
X = data_bj.drop('AQI指数', axis=1)
y = data_bj['AQI指数']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, shuffle=False)
models = {
    "Linear Regression": LinearRegression(),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42)
}

results = {}
for model_name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    results[model_name] = {"MSE": mse, "R²": r2}

# 打印结果
print("模型效果对比：")
for model_name, metrics in results.items():
    print(f"{model_name} - MSE: {metrics['MSE']:.4f}, R²: {metrics['R²']:.4f}")
```

```python
import matplotlib.pyplot as plt

predictions = {}
for model_name, model in models.items():
    y_pred = model.predict(X_test)
    predictions[model_name] = y_pred

plt.figure(figsize=(10, 6), dpi=200)
plt.plot(y_test.values, label='True Values', marker='o', linestyle='-', color='black', alpha=0.7)

for model_name, y_pred in predictions.items():
    plt.plot(y_pred, label=f'{model_name} Predictions', marker='x', linestyle='--', alpha=0.7)

plt.legend()
plt.title('AQI Prediction for BJ (Comparison of Models)')
plt.xlabel('Sample Index')
plt.ylabel('AQI')
plt.grid(False)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 模型名称和对应的指标
models = ['Linear Regression', 'Gradient Boosting', 'Random Forest']
mse_values = [16.2464, 5.7204, 6.1027]
r2_values = [0.9048, 0.9665, 0.9642]

# 设置柱状图位置和宽度
x = np.arange(len(models))
width = 0.35

# 创建双柱状图
fig, ax1 = plt.subplots(figsize=(8, 4), dpi=200)

# MSE 柱状图
bars1 = ax1.bar(x - width / 2, mse_values, width, label='MSE', color='skyblue')
ax1.set_ylabel('MSE', color='skyblue', fontsize=12)
ax1.tick_params(axis='y', labelcolor='skyblue')
ax1.set_xticks(x)
ax1.set_xticklabels(models)
ax1.set_title('Model Performance Comparison', fontsize=14)

# 添加 MSE 数值标签
for bar in bars1:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width() / 2, height, f'{height:.2f}', ha='center', va='bottom', fontsize=10)

# R² 柱状图
ax2 = ax1.twinx()
bars2 = ax2.bar(x + width / 2, r2_values, width, label='$R^2$', color='orange')
ax2.set_ylabel('$R^2$', color='orange', fontsize=12)
ax2.tick_params(axis='y', labelcolor='orange')

# 添加 R² 数值标签
for bar in bars2:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width() / 2, height, f'{height:.4f}', ha='center', va='bottom', fontsize=10)

# 图例
fig.tight_layout()
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')

plt.show()
```

```python
import pandas as pd

# 创建一个 DataFrame 保存真实值和每个模型的预测值
results_df = pd.DataFrame({'True Values': y_test})
for model_name, y_pred in predictions.items():
    results_df[model_name] = y_pred


plt.figure(figsize=(10, 3), dpi=200)
for i, model_name in enumerate(predictions.keys()):
    plt.subplot(1, 3, i + 1)
    plt.scatter(results_df['True Values'], results_df[model_name], alpha=0.7)
    plt.plot([results_df['True Values'].min(), results_df['True Values'].max()], 
             [results_df['True Values'].min(), results_df['True Values'].max()], 
             color='red', linestyle='--', linewidth=2)
    plt.title(f'{model_name}: True vs Predicted')
    plt.xlabel('True Values')
    plt.ylabel('Predicted Values')
    plt.grid()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 3), dpi=200)
for i, model_name in enumerate(predictions.keys()):
    residuals = results_df['True Values'] - results_df[model_name]
    plt.subplot(1, 3, i + 1)
    plt.hist(residuals, bins=20, alpha=0.7, color='green', edgecolor='black')
    plt.title(f'{model_name}: Residuals Distribution')
    plt.xlabel('Residuals')
    plt.ylabel('Frequency')
    plt.grid(False)

plt.tight_layout()
plt.show()
```
