```python
import time
import requests
from bs4 import BeautifulSoup

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 6.1; WOW64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/63.0.3239.132 Safari/537.36'}

for city in ['nanjing']:
    with open(f'air_{city}_data.csv', 'a+', encoding='utf-8-sig') as f:
        f.write('日期,质量等级,AQI指数,当天AQI排名,PM2.5,PM10,So2,No2,Co,O3\n')
        for y in range(2024, 2025):
            for i in range(1, 13):
                time.sleep(2)
                # 使用f-string格式化月份为两位数
                month_str = f"{i:02d}"
                try:
                    url = f'http://www.tianqihoubao.com/aqi/{city}-{y}{month_str}.html'
                    print(f'正在抓取{city}{y}年{i}月数据...')
                    response = requests.get(url=url, headers=headers, timeout=15, verify=False)
                    print(f"状态码：{response.status_code}")  # 打印状态码
                    if response.status_code != 200:
                        print(f"无法获取{y}年{i}月的数据，状态码：{response.status_code}")
                        continue
                    soup = BeautifulSoup(response.text, 'html.parser')
                    tr = soup.find_all('tr')
                    # 去除标签栏
                    for j in tr[1:]:
                        td = j.find_all('td')
                        if len(td) >= 10:  # 确保有足够的td元素
                            Date = td[0].get_text().strip()
                            Quality_level = td[1].get_text().strip()
                            AQI = td[2].get_text().strip()
                            AQI_rank = td[3].get_text().strip()
                            PM25 = td[4].get_text().strip()
                            PM10 = td[5].get_text().strip()
                            SO2 = td[6].get_text().strip()
                            NO2 = td[7].get_text().strip()
                            CO = td[8].get_text().strip()
                            O3 = td[9].get_text().strip()

                            f.write(f"{Date},{Quality_level},{AQI},{AQI_rank},{PM25},{PM10},{SO2},{NO2},{CO},{O3}\n")
                        else:
                            print(f"跳过行：td元素数量不足：{td}")
                except Exception as e:
                    print(f"抓取{y}年{i}月数据时出错：{e}")
```
